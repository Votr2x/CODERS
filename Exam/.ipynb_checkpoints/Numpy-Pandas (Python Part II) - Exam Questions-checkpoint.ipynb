### 3. If there are any issues with the data types in the DataFrame, try to resolve them.  
### If modification of the date/time data is required, perform it using a `lambda` function.

```python
for col in df.columns:
    if col == 'date':
        df['date'] = df['date'].apply(lambda x: pd.to_datetime(x, errors='coerce'))
    else:
        df[col] = df[col].astype(float)

print(df.dtypes)
print(df.head())
```

### 6. Verilən DataFrame-də outlier-ləri (anormal dəyərləri) tapmağa çalışın.  
#### - Outlier-lər olan sütunların adlarını çıxarın və bu sütunları boxplotlar vasitəsilə vizuallaşdırın.  
#### - Əgər outlier olan sütunlar məlumdursa, bu sütunlardakı dəyərləri 10000-ə vuraraq dəyişdirin və yenidən boxplot ilə vizuallaşdırın.

---

### 6. Attempt to identify the outliers in the given DataFrame.  
#### - Extract the names of the columns containing outliers and visualize them using boxplots.  
#### - If the columns with outliers are already known, multiply the values in these columns by 10000 and visualize the updated data again with boxplots.

```python
col = col = [i for i in df.columns if df[i].dtype == float]
l = []
for i in col:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    if ((df[i] < lower_bound) | (df[i] > upper_bound)).any():
        l.append(i)
print(l)
```

### 9. Yalnız `float64` tipli sütunların məlumat xüsusiyyətlərinə (data properties) baxın.  
#### - Əgər yeni bir sütunun miqyası (scale-i) fərqlidirsə, bu sütunlara normallaşdırma və ya standartlaşdırma üsulunu manual qaydada tətbiq edin.  
#### - Funksiyanı yazarkən `numpy` kitabxanasından istifadə edin.

---

### 9. Examine the data properties of the columns with a `float64` data type only.  
#### - If the scale of a new column differs, apply either normalization or standardization manually.  
#### - Use the `numpy` library when writing the function.

```python
def norm_stand(df, method):
    col = [i for i in df.columns if df[i].dtype == float]
    if method == 'norm':
        print((df[col] - df[col].min()) / (df[col].max() - df[col].min()))
    elif method == 'stand': 
        print((df[col] - df[col].mean()) / df[col].std())
norm_stand(df, 'norm')
norm_stand(df, 'stand')
```

### 10. Create a new column consisting of at least 10 letters.  
#### - Name the column **Label**.  
#### - Apply a labeling method to replace the letters in this column with numbers (use a function).  
#### - Convert the result of the labeling into a `pandas DataFrame` and merge it with the main DataFrame.

```python
from sklearn.preprocessing import LabelEncoder
let = ['a', 'b', 'c', 'd', 'e', 'g', 'h', 'i', 'j', 'k']
df['Label'] = np.zeros(len(df))
for i in range(len(df['Label'])):
    df['Label'].iloc[i] = np.random.choice(let)
encoder = LabelEncoder()
df['Label'] = encoder.fit_transform(df['Label'])
print(df['Label'])
```
