```python
import pandas as pd
import numpy as np

data = {
    'Product': ['Apple', 'Banana', 'Orange', 'Grapes', 'Mango', 'Banana', 'Apple'],
    'Category': ['Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit'],
    'Price': [120, 80, 100, 90, 150, 85, 130],
    'Quantity': [10, 5, 7, 3, 12, 8, 6],
    'Country': ['Azerbaijan', 'Turkey', 'India', 'Iran', 'Azerbaijan', 'Turkey', 'India']
}

df = pd.DataFrame(data)
df
```

```python
#1. Product sütununu Label Encoding və One-Hot Encoding ilə kodlaşdır.
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['EncoderProduct'] = encoder.fit_transform(df['Product'])
print(df)
```

```python
df_encoded = pd.get_dummies(df, columns=['Product'])
print(df_encoded)
```

```python
#2.Country sütununu LabelEncoder istifadə edərək rəqəmlərə çevir.
```

```python
df['EncoderCountry'] = encoder.fit_transform(df['Country'])
print(df)
```

```python
#3.Price sütununu standartlaşdır (z-score normalization) və yeni sütun əlavə et.
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df['Standart'] = scaler.fit_transform(df[['Price']])
print(df)
```

```python
#4.Quantity sütununu min-max normalization ilə [0,1] aralığına gətir.
```

```python
from sklearn.preprocessing import MinMaxScaler
minmaxscaler = MinMaxScaler()
df['MinMaxScaled'] = minmaxscaler.fit_transform(df[['Quantity']])
print(df)
```

```python
#5.Hər Country üzrə məhsulların orta qiymətini tap.
```

```python
avgpr = df.groupby(['Country'])['Price'].mean()
print(avgpr)
```

```python
#6.Ən çox satılan məhsulu (ümumi Quantity-yə görə) tap.
```

```python
maxpr = df.groupby(['Product'])['Quantity'].max()
print(maxpr)
```

```python
#7.Price və Quantity sütunlarını NumPy array-ə çevir və element-wise hasillərini tap (Price * Quantity).
```

```python
import numpy as np
pricearr = np.array(df['Price'])
quantityarr = np.array(df['Quantity'])
print(pricearr * quantityarr)
```

```python
#8.Yuxarıdakı hasillərdən istifadə edərək ümumi gəlir (revenue) hesabla.
```

```python
tot = pricearr * quantityarr
print(tot.sum())
```
