```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'Name_Age': ['Alice_25', 'Bob_30', 'Charlie_35', 'David_40', 'Eva_28'],
    'Salary_Dept': ['50000_IT', '60000_HR', '55000_Finance', '70000_IT', '48000_HR'],
    'Experience_Years': [2, 5, 7, 10, 3],
    'Bonus': [5000, 6000, 5500, 7000, 4800],
    'City': ['Baku', 'Istanbul', 'London', 'Paris', 'Berlin'],
    'Performance_Score': ['A', 'B', 'A', 'C', 'B'],
    'Remote': ['Yes', 'No', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)
df
```

```python
# 1.Name_Age və Salary_Dept sütunlarını ayır: Name, Age, Salary, Department.
df[['Name', 'Age']] = df['Name_Age'].str.split('_', expand=True)
df[['Salary', 'Dept']] = df['Salary_Dept'].str.split('_', expand=True)
print(df)
```

```python
# 2.Age, Salary, Experience_Years, Bonus → int tipinə çevir
# Convert relevant columns to int
df[['Age', 'Salary', 'Experience_Years', 'Bonus']] = df[['Age', 'Salary', 'Experience_Years', 'Bonus']].astype(int)
print(df)
```

```python
# 3.Performance_Score → Label Encoding tətbiq et.
encoder = LabelEncoder()
df['EncoderPerformance'] = encoder.fit_transform(df['Performance_Score'])
print(df)
```

```python
# 4.Remote və City → One-Hot Encoding tətbiq et.
df_encoded = pd.get_dummies(df, columns=['City'])
print(df_encoded)
```

```python
# 5.Analiz:
# Ən yüksək maaş alan şəxs
# Hər Department üzrə orta Salary və orta Experience_Years
# Hər Performance_Score üzrə orta Bonus
```

```python
print(df['Salary'].max())
print(df.groupby(['Dept'])['Salary'].mean())
print(df.groupby(['Dept'])['Experience_Years'].mean())
```

```python
# 6.Experience_Years və Age-dən istifadə edərək Seniority_Level sütunu yarat.
# Age < 30 → Junior
# Age 30–35 → Mid
# Age > 35 → Senior
# Create Seniority_Level based on Age
df['Seniority_Level'] = df['Age'].apply(lambda x: 'Junior' if x < 30 else ('Mid' if x <= 35 else 'Senior'))
print(df)
```

```python
# 7. Salary sutununa Normalization ve Standardization tetbiq edin.
minmax = MinMaxScaler()
df['MinMax'] = minmax.fit_transform(df[['Salary']])
standatd = StandardScaler()
df['StandardScaled'] = standatd.fit_transform(df[['Salary']])
print(df)
```
