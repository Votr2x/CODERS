```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Product': ['Apple', 'Banana', np.nan, 'Orange', 'Grapes'],
    'Price': [100, 200, 150, None, 250],
    'Quantity': [10, None, 15, 20, 5],
    'Category': ['Fruit', 'Fruit', 'Fruit', 'Fruit', None],
    'Discount': [10, -5, 20, 15, 0]
}
data
```

```python
#1. Sətirlərdə və sütunlarda olan NaN dəyərləri təmizlə.
```

```python
df = pd.DataFrame(data)
df_new = df.dropna().copy()
print(df_new)
```

```python
#2. Sütun adlarını dəyişdirərək onlara yeni adlar vermək:
# Sütun adlarını dəyişdirmək
```

```python
df.columns = ['Product1', 'Price1', 'Quantity1', 'Category1', 'Discount1']
print(df)
```

```python
#3. Discount sütununda mənfi dəyərləri tap və onları sıfırla əvəz et.
```

```python
df.loc[df['Discount1'] < 0, 'Discount1'] = 0
print(df)
```

```python
#4.Price və Quantity sütunlarının hasilini hesabla və yeni bir Total_Sales sütunu əlavə et.
```

```python
df['Total_Sales'] = df['Price1'] * df['Quantity1']
print(df)
```

```python
#5.Category sütununda boş dəyərləri "Unknown" ilə doldur.
```

```python
df['Category1'] = df['Category1'].fillna('Unknown')
print(df)
```

```python
#6.Boş olan Product dəyərlərini "Unnamed Product" ilə doldur.
```

```python
df['Product1'] = df['Product1'].fillna('Unnamed Product')
print(df)
```

```python
#7.Qiyməti olmayan (Price) sətrləri sil.
```

```python
df = df.dropna(subset=['Price1'])
print(df)
```
