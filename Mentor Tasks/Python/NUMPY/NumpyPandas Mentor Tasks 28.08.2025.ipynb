```python
import pandas as pd

data = {
    "Title": ["Inception", "Titanic", "Avengers", "Friends", "Parasite", "Breaking Bad", "Interstellar", "Joker"],
    "Genre": ["Sci-Fi", "Romance", "Action", "Comedy", "Thriller", "Crime", "Sci-Fi", "Drama"],
    "Country": ["USA", "USA", "USA", "USA", "South Korea", "USA", "USA", "USA"],
    "Duration": [148, 195, 143, 22, 132, 47, 169, 122],  # dəqiqə
    "Rating": [8.8, 7.8, 8.0, 8.9, 8.6, 9.5, 8.6, 8.4],
    "Type": ["Movie", "Movie", "Movie", "Series", "Movie", "Series", "Movie", "Movie"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# 1.Rating ≥ 8.5 ve Duration > 150 dəqiqə olan filmləri seçin.
```

```python
df[(df['Rating'] >= 8.5) & (df['Duration'] >= 150)]
```

```python
# 2 .Ən qısa və ən uzun filmi seçin.Rating sütununa görə ise azalan sıraya düzün.
```

```python
movies = df[df['Type'] == 'Movie']
shortest = movies.loc[movies['Duration'].idxmin()]
longest = movies.loc[movies['Duration'].idxmax()]
new_df = pd.DataFrame([shortest, longest])
new_df = new_df.sort_values(by = 'Rating', ascending = False)
print(new_df)
```

```python
# 3.Type sütunu üzrə ortalama Ratingi,Country sütunu üzrə ortalama Duration tapın.
```

```python
avg_rating_by_type = df.groupby("Type")["Rating"].mean()
avg_duration_by_country = df.groupby("Country")["Duration"].mean()
print(avg_rating_by_type)
print(avg_duration_by_country)
```

```python
# 4.Duration_hours = Duration / 60 sütununu əlavə edin.High_Rating sütunu yaradın: Rating ≥ 8.5 → 1, əks halda 0.
```

```python
df["Duration_hours"] = df["Duration"] / 60
df["High_Rating"] = (df["Rating"] >= 8.5).astype(int)
print(df)
```

```python
# 5.Genre, Country, Type sütunlarını LabelEncoder ilə kodlaşdırın.Yeni kodlaşdırılmış sütunları DataFrame-ə əlavə edin.
```

```python
# 6.Rating və Duration sütunlarını standartlaşdırın (Z-score).Orta və standart sapmanı hesablayın.
```

```python
# Calculate mean and std for Rating and Duration
rating_mean = df["Rating"].mean()
rating_std = df["Rating"].std()
duration_mean = df["Duration"].mean()
duration_std = df["Duration"].std()
df["Rating_z"] = (df["Rating"] - rating_mean) / rating_std
df["Duration_z"] = (df["Duration"] - duration_mean) / duration_std
print(df)
```

```python
# 7.Type = "Movie" olan filmlərin Genre üzrə ortalama Rating-ini tapın.
# Duration_hours > 2 olan filmləri seçin.
```

```python
avg_rating_by_genre = df[df["Type"] == "Movie"].groupby("Genre")["Rating"].mean()
print(avg_rating_by_genre)
long_movies = df[(df["Type"] == "Movie") & (df["Duration"] > 120)]
print(long_movies)
```

```python
# 8.Verilən DataFrame-də sütunların seyrək (sparse) və ya tam (complete) olub-olmadığını analiz edin.
```

```python
sparsity = df.isna().any().replace({True: "Sparse", False: "Complete"})
print(sparsity)
```
