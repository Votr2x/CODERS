```python
sides = [1, 2, 3, 4, 5, 6]
even = len([i for i in sides if (i % 2) == 0])
tot = len(sides)
prob = even / tot
print(prob)
```

```python
# 2. Aşağıdakı verilmiş məlumatlar üçün mean, median və mode-u hesablayın:
# Verilmiş məlumatlar: [12, 15, 14, 10, 12, 12, 18, 16, 14, 14]
```

```python
l = [12, 15, 14, 10, 12, 12, 18, 16, 14, 14]
l_sorted = sorted(l)
mean = sum(l_sorted) / len(l_sorted)
print(f' Mean: {mean}')
if (len(l_sorted) / 2) == 1:
    ind_median = ((len(l_sorted) / 2) + 1)
    print(f' Median: {l_sorted[ind_median]}')
else:
    ind_median = len(l_sorted) // 2
    median = (l_sorted[ind_median - 1] + l_sorted[ind_median]) / 2
    print(f' Median: {median}')
max_num = max(l.count(x) for x in l)
mode = [x for x in l if l.count(x) == max_num]
mode = list(set(mode))
print(f' Mode: {mode}')
```
