```python
# 1.Bir şirkətdə işçilərin orta maaşının 2500 AZN olduğu iddia olunur.30 nəfərdən ibarət nümunə götürülüb.
# orta maaş 2600, standart sapma isə 300-dür.
# a) Bu nəticəyə əsasən şirkətdə maaşın 2500-dən fərqli olub-olmadığını t-test ilə yoxlayın. b) p-value şərh edin.
```

```python
import numpy as np
from scipy.stats import ttest_ind
np.random.seed(42)
arr = np.random.normal(loc = 2600, scale = 300, size = 30)
t, p = ttest_ind(arr, 2500)
print(p)
```

```python
# 2. Bir dərsdə qızların və oğlanların orta imtahan nəticələri müqayisə edilir:
# Qızlar: [78, 85, 82, 90, 76, 88, 84] Oğlanlar: [72, 70, 68, 75, 80, 74, 73]
# a) İki qrup arasında əhəmiyyətli fərq olub-olmadığını yoxlayın (independent t-test). b) Nəticəni şərh edin.
```

```python
girls = [78, 85, 82, 90, 76, 88, 84]
boys = [72, 70, 68, 75, 80, 74, 73]
ttest_ind(girls, boys)
```

```python
# 3. Oğlan və qız tələbələrin riyaziyyat balları arasında fərq varmı?
# 50 oğlan və 50 qız üçün ballar generasiya et və independent t-test apar. α = 0.05 ilə nəticəni şərh et.
```

```python
girls = np.array([78, 85, 82, 90, 76, 88, 84])
boys = np.array([72, 70, 68, 75, 80, 74, 73])
bb = np.random.normal(loc = boys.mean(), scale = boys.var(), size = 50)
gg = np.random.normal(loc = girls.mean(), scale = girls.var(), size = 50)
ttest_ind(bb, gg)
```

```python
# 4. Şirkət iddia edir ki, ortalama maaş 1200 AZN-dir. 40 işçinin maaşı verilib — bu iddianı yoxla.
# 40 maaş nümunəsi generasiya et (ortalama 1150, sd 150) və one-sample t-test apar.
```

```python
from scipy.stats import ttest_1samp
arr = np.random.normal(loc = 1150, scale = 150, size = 40)
t, p = ttest_1samp(arr, 1200)
print(p)
```

```python
# 5. 200 nəfərlik nümunədə kişilər və qadınlar, və siqaretçi/siqaret çəkən deyil məlumatı ilə chi-square testi apar.
# Cinsiyyət ilə siqaret çəkmə arasında əlaqə varmı?
```

```python
from scipy.stats import chi2_contingency
import pandas as pd
genders = np.random.choice(["Male", "Female"], size = 200)
smoking = np.random.choice(["Smoker", "Non-smoker"], size = 200)
crosstab = pd.crosstab(genders, smoking)
chi2_contingency(crosstab)
```

```python
# 6. Ortalama 70, sd 10 olan paylanmada tələbənin 60-dan yuxarı bal toplama ehtimalı neçədir? Normal CDF istifadə et.
```

```python
from scipy.stats import norm
print(f'Prob: {1 - norm.cdf(60, 70, 10)}')
```

```python
# 7. Bir fabrikdə bir saat ərzində istehsal edilən qüsurlu məhsulların sayı Poisson paylanmasına əsasən modelləşdirilir.
# Orta hesabla hər saatda 3 qüsurlu məhsul olur. Bir saat ərzində heç bir qüsurlu məhsulun olmaması ehtimalını hesablayın.
```

```python
from scipy.stats import poisson
print(f'Poisson prob: {poisson.pmf(0, 3)}')
```

```python
# 8.Gündə ortalama 12 müştəri gəlir. Sabah 15 müştərinin gəlmə ehtimalı neçədir? Poisson PMF istifadə et.
```

```python
print(f'Poisson prob: {poisson.pmf(15, 12)}')
```

```python
# 9. 25 tələbənin əvvəl və sonra ballarını generasiya et və paired t-test apar.
# Yeni tədris metodundan əvvəl və sonra eyni tələbələrin balları verilib — metod effektivdirmi?
```

```python
from scipy.stats import ttest_rel
before_scores = np.random.normal(loc=70, scale=10, size=25)
after_scores = before_scores + np.random.normal(loc=5, scale=5, size=25)
ttest_rel(before_scores, after_scores)
```
