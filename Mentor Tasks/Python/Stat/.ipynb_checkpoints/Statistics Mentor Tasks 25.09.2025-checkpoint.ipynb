```python
#1. 50 nəfər tələbənin gündəlik oxuma saatlarını təsadüfi olaraq (Normal paylanma ilə) generasiya et.
# Onların ortalamasını, medianını, modunu, dispersiyasını və standart sapmasını hesabla.
```

```python
import numpy as np
import pandas as pd
arr_np = np.random.normal(size = 50)
arr = pd.Series(arr_np)
print(f'Mean: {arr.mean()}')
print(f'Median: {arr.median()}')
print(f'Mode: {arr.mode()[0]}')
print(f'Var: {arr.var()}')
print(f'Std: {arr.std()}')
```

```python
#2. Bir şirkətdə qadın və kişi işçilərin maaşları arasında fərq olub-olmadığını yoxla.
# 40 qadın, 40 kişi üçün maaş datasetini yarat (Normal paylanma əsasında).T-test ilə analiz et və nəticəni izah et.
```

```python
from scipy.stats import ttest_ind
np.random.seed(42)
df = pd.DataFrame()
man = np.random.normal(loc = 1000, size = 40)
woman = np.random.normal(loc = 1000, size = 40)
df['Man'] = man
df['Woman'] = woman
ttest_ind(df['Man'], df['Woman'])
```

```python
#3. Bir mağazada müştəri məmnuniyyəti sorğusu aparılıb: Çox razı, Razı, Narazı.
# Qadın və kişi müştərilərin cavablarını təsadüfi generasiya et.
# Chi-Square testi ilə cinsiyyət və məmnuniyyət arasında əlaqə olub-olmadığını yoxla.|
```

```python
from scipy.stats import chi2_contingency
genders = np.random.choice(["Male", "Female"], size = 200)
rate = np.random.choice(["Very Happy", "Happy", "Unhappy"], size = 200)
crosstab = pd.crosstab(genders, rate)
chi2_contingency(crosstab)
```

```python
#4. fərqli tədris metodunun tələbələrin nəticəsinə təsirini yoxla.
# 30 tələbəlik hər qrup üçün balları generasiya et.
# ANOVA testi apar və qruplar arasında fərq olub-olmadığını təhlil et.
```

```python
from scipy.stats import f_oneway
group1 = np.random.randint(50, 101, 30)
group2 = np.random.randint(50, 101, 30)
group3 = np.random.randint(50, 101, 30)
f_oneway(group1, group2, group3)
```

```python
#5. Bir fabrikdə saatda orta hesabla 5 qüsurlu məhsul istehsal olunur.Poisson paylanması ilə hesabla:
# Bir saatda heç qüsurlu məhsul olmaması ehtimalı.
# Ən azı 3 qüsurlu məhsul olması ehtimalı.
```

```python
from scipy.stats import poisson
print(f'Prob 0: {poisson.pmf(0, 5)}')
print(f'Prob min 3: {1 - poisson.cdf(2, 5)}')
```

```python
#6. 20 tələbənin yeni tədris metodundan əvvəl və sonra topladığı balları müqayisə et.
# Paired t-test apar və metodun effektiv olub-olmadığını yoxla.
```

```python
from scipy.stats import ttest_rel
before_scores = np.random.normal(loc=70, scale=10, size=20)
after_scores = before_scores + np.random.normal(loc=5, scale=5, size=20)
ttest_rel(before_scores, after_scores)
```
