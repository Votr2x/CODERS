```python
# 2. Aşağıdakı verilmiş məlumatlar üçün mean, median və mode-u hesablayın:
# Verilmiş məlumatlar: [12, 15, 14, 10, 12, 12, 18, 16, 14, 14]
```

```python
l = [12, 15, 14, 10, 12, 12, 18, 16, 14, 14]
l_sorted = sorted(l)
mean = sum(l_sorted) / len(l_sorted)
print(f' Mean: {mean}')
if (len(l_sorted) / 2) == 1:
    ind_median = ((len(l_sorted) / 2) + 1)
    print(f' Median: {l_sorted[ind_median]}')
else:
    ind_median = len(l_sorted) // 2
    median = (l_sorted[ind_median - 1] + l_sorted[ind_median]) / 2
    print(f' Median: {median}')
max_num = max(l.count(x) for x in l)
mode = [x for x in l if l.count(x) == max_num]
mode = list(set(mode))
print(f' Mode: {mode}')
```

```python
# 3. Əgər bir tərs-mükəmməl (bias) coin 70% ehtimalla şəkil tərəfi gələrsə,
# 5 dəfə atıldıqda 3 dəfə şəkil gəlmə ehtimalını hesablayın.
```

```python
from math import comb
probability_head = 0.7
total_tries = 5
head_result = 3
final_prob = comb(total_tries, head_result) * (probability_head ** head_result) * ((1 - probability_head)**(total_tries-head_result))
print(final_prob)
```

```python
# 4. Aşağıdakı verilmiş məlumatlar üçün variance və standart sapmanı hesablayın:
# Verilmiş məlumatlar: [8, 10, 12, 14, 16, 18, 20]
```

```python
import numpy as np
arr = np.array([8, 10, 12, 14, 16, 18, 20])
std = np.std(arr)
var = np.var(arr)
print(std, var)
```

```python
# 5. Mean 50 və std-si 10 olan normal distribution malik 1000 məlumat
# nöqtəsi yaradın və histogramını çəkin.
```

```python
data = np.random.normal(50, 10, 1000)
```

```python
import matplotlib.pyplot as plt
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Histogram of Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
