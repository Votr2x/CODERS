```python
# 4. Bir fabrikdə bir saat ərzində istehsal edilən qüsurlu məhsulların sayı Poisson paylanmasına əsasən modelləşdirilir.
# Orta hesabla hər saatda 3 qüsurlu məhsul olur. Bir saat ərzində heç bir qüsurlu məhsulun olmaması ehtimalını hesablayın.
```

```python
import math
defect = 0
avg_def = 3
prob = math.exp(-avg_def) * (avg_def ** defect) / math.factorial(defect)
print(prob)
```

```python
# 8. Verilən datasetdə ekstremal qiymətləri (outliers) müəyyən edin:
# Dataset: [12, 15, 14, 10, 20, 25, 415, 30, 22, 626, 35, 40, 398, 42, 5]
# Ekstremal qiymətləri aşkar etmək üçün 1-ci kvartil (Q1) və 3-cü kvartil (Q3) əsasında interquartile range (IQR)
# metodundan istifadə edin.
```

```python
data = [12, 15, 14, 10, 20, 25, 415, 30, 22, 626, 35, 40, 398, 42, 5]
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR
outliers = [x for x in data if x < lower_fence or x > upper_fence]
print(outliers)
```

```python
# 9. Mean 50 və std-si 10 olan normal paylanmadan təsadüfi 1000 məlumat yaradın.
# Verilənlər arasında std-dən 2 dəfə böyük outlier-ləri müəyyən edin.
```

```python
import numpy as np
mean = 50
std_dev = 10
n_points = 1000
data = np.random.normal(mean, std_dev, n_points)
lower_bound = mean - 2*std_dev
upper_bound = mean + 2*std_dev
outliers = data[(data < lower_bound) | (data > upper_bound)]
print(outliers)
```

```python
# 10. Bir qutuda 5 qırmızı, 4 yaşıl və 3 mavi top var. Birinci top çıxarıldıqdan sonra geri qoymadan ikinci top çəkilir.
# Çıxarılma ardıcıllığı fərqli rənglərə aid olarsa (məsələn, əvvəl qırmızı, sonra yaşıl),
# həmin ardıcıllığın ehtimalını hesablayan funksiya yazın.
# Funksiya istifadəçinin daxil etdiyi rəng ardıcıllığına əsasən ehtimalı qaytarmalıdır.

# Məsələn: calculate_probability(["red", "green"])
```

```python
def calculate_probability(*args):
    colors = list(args)
    box = ['red'] * 5 + ['green'] * 4 + ['blue'] * 3
    length = len(box)
    prob = 1
    for i in colors:
        if i == 'red':
            prob *= (box.count('red') / length)
            box.remove('red')
        elif i == 'green':
            prob *= (box.count('green') / length)
            box.remove('green')
        elif i == 'blue':
            prob *= (box.count('blue') / length)
            box.remove('blue')
    print(prob)
calculate_probability('red', 'green')
```
