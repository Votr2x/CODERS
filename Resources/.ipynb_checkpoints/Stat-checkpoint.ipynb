### Theory of Probabilities basics

If two events cannot occur at the same time we use Addition Theorem

```python
#If independent
```

```python
# Event A: Rolling an even number on a die
# Event B: Rolling a number bigger that 3 on a die
prob_A = 3/6
prob_B = 3/6
# Probability of event A or event B occuring is: 
prob_A_and_B = 2/6
prob_A_or_B = prob_A + prob_B - prob_A_and_B
print(prob_A_or_B)
```

```python
#If dependent
```

```python
# Event A: Tossing a head on a coin
# Event B: Tossing a tail on a coin
prob_A = 1/2
prob_B = 1/2
# Probability of event A or event B occuring is:
prob_A_or_B = prob_A + prob_B
print(prob_A_or_B)
```

If two events can occur at the same time

```python
#If dependent
```

```python
# Event A: Taking a red ace from a card deck
# Event B: Taking a red card after taking a red ace
prob_A = 2/52
prob_B = (26 - 1)/(52 - 1)
# Probability of event A and event B occuring is:
prob_A_and_B = prob_A * prob_B
print(prob_A_and_B)
```

```python
#If independent
```

```python
# Event A: Tossing a head on a coin
# Event B: Rolling an even number on a dime
prob_A = 1/2
prob_B = 3/6
# Probability of event A and event B occuring is:
prob_A_and_B = prob_A * prob_B
print(prob_A_and_B)
```

### Statistics basics

```python
# Functions
import pandas as pd
df = pd.Series([1, 2, 3, 3, 4, 5])
print(f'Mode: {df.mode()}')
print(f'Median: {df.median()}')
print(f'Mean: {df.mean()}')
print(f'Var: {df.var()}')
print(f'Std: {df.std()}')
```
