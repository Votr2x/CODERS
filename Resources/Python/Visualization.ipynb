```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
        np.random.randn(1000, 4),  #Creates matrix with 1000 rows and 4 columns in std
        index=pd.date_range('1/1/2000', periods=1000), #Indexes of the dataset are dates, starting from 1/1/2000, 1000 days
        columns=list('ABCD') #Name of 4 columns
                 )
row_6 = df.iloc[5] #Find 6th row
row_6.plot(kind='bar', rot=45) #Type of plot is bar and rotation of column names (x axis) is 4 degrees
plt.axhline(0, color='r') #Creating an line on y = 0 with red color
plt.show()
```
