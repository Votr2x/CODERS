```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
        np.random.randn(1000, 4),  #Creates matrix with 1000 rows and 4 columns in std
        index=pd.date_range('1/1/2000', periods=1000), #Indexes of the dataset are dates, starting from 1/1/2000, 1000 days
        columns=list('ABCD') #Name of 4 columns
                 )
row_6 = df.iloc[5] #Find 6th row
row_6.plot(kind='bar', rot=45) #Type of plot is bar and rotation of column names (x axis) is 4 degrees
plt.axhline(0, color='r') #Creating an line on y = 0 with red color
plt.show()
```

```python
df2 = pd.DataFrame(np.random.rand(10, 4), columns=['a', 'b', 'c', 'd'])
df2.plot(kind='bar', rot=90);
```

```python
df2 = pd.DataFrame(np.random.rand(10, 4), columns=['a', 'b', 'c', 'd'])
df2.plot(kind='bar')
plt.xticks(rotation=45)  # Adjust the rotation angle as needed
plt.yticks(rotation=45)  # Adjust the rotation angle as needed
plt.show()
```

```python
df2.plot(kind='bar', stacked=True, rot=0) #Instead of giving side by side gives stacked
```
