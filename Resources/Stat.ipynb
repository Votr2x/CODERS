```python
# Types of probabilities
# 1) Classical = rolling 3 on a dime(1 / 6)
# 2) Empirical = probability based on observations = flipping 100 times a coin gave us 60 heads(60 / 100 prob of getting heads)
# 3) Subjective = probability based on personal judjment
```

```python
# Types of events
# 1) Simple event = consists only of 1 outcome
# 2) Compound event = consists of 2 or more outcomes
# 3) Independent events = one event's outcome does not affect other event's probability
# 4) Dependent events = one event's outcome affect other event's probability
# 5) Compatible events = event outcomes can occur simultaneously(Event A = rolling a dime will give us an even number, Event B = rolling a dime will give us a number > 3)
# 6) Incompatible events = event outcomes cannot occur simultaneously(Event A = rolling a dime will give us 5, Event B = roling a dime will give us 2)
```
