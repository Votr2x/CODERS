### Theory of Probabilities basics

If two events cannot occur at the same time we use Addition Theorem

```python
#If independent
```

```python
# Event A: Rolling an even number on a die
# Event B: Rolling a number bigger that 3 on a die
prob_A = 3/6
prob_B = 3/6
# Probability of event A or event B occuring is: 
prob_A_and_B = 2/6
prob_A_or_B = prob_A + prob_B - prob_A_and_B
print(prob_A_or_B)
```

```python
#If dependent
```

```python
# Event A: Tossing a head on a coin
# Event B: Tossing a tail on a coin
prob_A = 1/2
prob_B = 1/2
# Probability of event A or event B occuring is:
prob_A_or_B = prob_A + prob_B
print(prob_A_or_B)
```

### Statistics basics

```python
# Functions
import pandas as pd
df = pd.Series([1, 2, 3, 3, 4, 5])
print(f'Mode: {df.mode()}')
print(f'Median: {df.median()}')
print(f'Mean: {df.mean()}')
print(f'Var: {df.var()}')
print(f'Std: {df.std()}')
```

```python
# Defining outliers
# Outliers = Data points that differ significantly from most of the data
import pandas as pd
df = pd.DataFrame({'A': [10, 12, 11, 13, 12, 110]})
Q1 = df['A'].quantile(0.25)
Q3 = df['A'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['A'] < lower) | (df['A'] > upper)]
print(outliers)
```

```python
# Another method using z_scores
from scipy import stats
import numpy as np

data = np.array([10, 12, 11, 13, 12, 110])
z_scores = stats.zscore(data)
outliers = data[np.abs(z_scores) > 2]
print(outliers)
```

```python
# Skewness is the symmetry of the curve. Ideal value is 0
import pandas as pd
data = pd.Series([10, 11, 12, 13, 10, 11, 14])
print("Skewness:", data.skew())
```

```python
# Kurtosis is how spread out is the data overall. if x = 0 then normal distribution, if x > 0 then peaked center, if x < 0 then flatter center
import pandas as pd
data = pd.Series([10, 11, 12, 13, 10, 11, 14])
print("Kurtosis:", data.kurt()) 
```
