Probability is a branch of mathematics that deals with the likelihood of events occurring. It provides a framework for quantifying uncertainty. In probability theory, an event is any outcome or set of outcomes of an experiment.

### Key Concepts:

- **Sample Space (S)**: The set of all possible outcomes of an experiment. It is denoted by $S$.
- **Event (E)**: A subset of the sample space. An event is a single outcome or a combination of outcomes.
- **Probability (P)**: A numerical measure of the likelihood that an event will occur, ranging from 0 to 1. $$0 \leq P(E) \leq 1.$$
- **Probability of an Event**: The probability of an event $E$, denoted by $P(E)$, is calculated as the ratio of the number of favorable outcomes to the total number of outcomes in the sample space.

### Example:

Consider rolling a fair six-sided die. Let's define the sample space $S$ and calculate the probability of certain events:

- **Sample Space**: $S = \{1, 2, 3, 4, 5, 6\}$
- **Probability of Rolling a 3 (Event E)**:
  - Number of favorable outcomes $(E)$ = 1 (rolling a 3)
  - Total number of outcomes $(S)$ = 6 (six-sided die)
  - $P(E) = \frac{1}{6}$

### Types of Probability

There are different types of probability that are used in various contexts. Let's explore each type:

#### 1. Classical Probability

Classical probability, also known as theoretical probability, is based on the assumption that all outcomes in a sample space are equally likely to occur. It is calculated by dividing the number of favorable outcomes by the total number of possible outcomes.

#### Example:

Consider rolling a fair six-sided die. The classical probability of rolling a 3 is calculated as:

$ P(\text{Rolling a 3}) = \frac{1}{6} $

#### 2. Empirical (or Experimental) Probability

Empirical probability is based on observed outcomes from an experiment or real-world data. It is calculated by dividing the number of times an event occurs by the total number of trials or observations.

##### Example:

Tossing a biased coin 100 times resulted in 60 heads. The empirical probability of getting heads when tossing the coin is:

$ P(\text{Heads}) = \frac{60}{100} = 0.6 $

#### 3. Subjective Probability

Subjective probability is based on personal judgment or belief about the likelihood of an event occurring. It is not based on objective data or mathematical calculations.

##### Example:

An individual predicts the likelihood of rain tomorrow based on their personal judgment and past experiences. They assign a subjective probability of 0.7 for rain.

### Types of Events

In probability theory, an event is a subset of the sample space, which represents the set of all possible outcomes of an experiment. Events can vary in complexity and relationship to each other. Let's explore the different types of events:

#### 1. Simple Event

A simple event is an event that consists of a single outcome in the sample space.

##### Example:

Rolling a fair six-sided die and getting a 3 is a simple event.

#### 2. Compound Event

A compound event is an event that consists of two or more outcomes in the sample space.

##### Example:

Tossing a fair coin and rolling a fair six-sided die are independent events. The compound event is the combination of these two events.

#### 3. Independent Events

Independent events are events where the occurrence of one event does not affect the probability of the other event.

##### Example:

Tossing a fair coin twice are independent events. The outcome of the first toss does not influence the outcome of the second toss.

#### 4. Dependent Events

Dependent events are events where the outcome of one event affects the probability of the other event.

##### Example:

Drawing two cards without replacement from a deck of cards is a dependent event. The probability of drawing a certain card on the second draw depends on the outcome of the first draw.

In probability theory, compatible(related) and incompatible(unrelated) events describe the relationship between two or more events regarding their possibility of occurring simultaneously.

#### Compatible(Related) Events
Compatible events (or intersecting events) are events that can occur simultaneously. That is, their intersection is not empty.

##### Examples of Compatible Events:
- **Rolling a Die**: Event A — an even number is rolled, and Event B — a number greater than 3 is rolled. These events are compatible because there are numbers (e.g., 4 and 6) that satisfy both conditions.

##### Mathematical Definition:
If events A and B are compatible, then:
$ P(A \cap B) \neq 0 $

where $ P(A \cap B) $ is the probability that both events A and B occur simultaneously.

#### Incompatible(Unrelated) Events
Incompatible events (or mutually exclusive events) are events that cannot occur simultaneously. That is, if one event occurs, the other necessarily does not occur.

##### Examples of Incompatible Events:
- **Flipping a Coin**: Event A — heads is flipped, and Event B — tails is flipped. These events are incompatible because they cannot occur simultaneously.
- **Rolling a Die**: Event A — a 2 is rolled, and Event B — a 5 is rolled. These events are incompatible because they cannot occur simultaneously.

##### Mathematical Definition:
If events A and B are incompatible, then:
$ P(A \cap B) = 0 $

where $ P(A \cap B) $ is the probability that both events A and B occur simultaneously.

#### Important Notes:
- **Compatibility and Independence**: Compatible events can be either dependent or independent. Independence of events means that the probability of one event does not depend on whether the other event has occurred, while compatibility merely indicates the possibility of their simultaneous occurrence.
- **Examples of Dependent Compatible Events**: Drawing a card from a deck without replacement. Event A — drawing an Ace, Event B — drawing a red card. These events are compatible and dependent because drawing one card changes the composition of the deck for the next draw.

Understanding the differences between compatible and incompatible events is crucial for the correct application of addition and multiplication rules in probability theory.

### Python Implementation:

```python
import random

# Defining the sample space for a six-sided die
sample_space = [1, 2, 3, 4, 5, 6]

# Defining a function to calculate probability
def calculate_probability(event, num_trials):
    favorable_outcomes = event.count(True)
    print(favorable_outcomes)
    print(num_trials)
    probability = (favorable_outcomes / num_trials)*100
    return probability

# Rolling the die 1000 times and calculating the probability of rolling a 3
num_trials = 1000
event = [random.choice(sample_space) == 3 for _ in range(num_trials)]

# Calculating probability
probability_of_3 = calculate_probability(event, num_trials)
print("Probability of rolling a 3:", probability_of_3)
```

### Addition and Multiplication Theorems in Probability Theory

#### Addition Theorem

1. **For Compatible Events:**
   - If events A and B are compatible (i.e., they can occur simultaneously), then:
     $$
     P(A \cup B) = P(A) + P(B) - P(A \cap B)
     $$
   - This formula applies to both dependent and independent compatible events.

2. **For Mutually Exclusive Events:**
   - If events A and B are mutually exclusive (i.e., they cannot occur simultaneously), then:
    $$
     P(A \cup B) = P(A) + P(B)
    $$
   - Since $ P(A \cap B) = 0 $, the intersection term is absent.
   - This formula applies to both dependent and independent compatible events.

#### Multiplication Theorem

1. **For Depented Compatible Events:**
   - If events A and B are compatible, the probability of their intersection is calculated using the general formula for the intersection of two events:
     $$
     P(A \cap B) = P(A) \cdot P(B|A)
     $$
   - Here, $ P(B|A) $ is the conditional probability of event B occurring given that event A has already occurred.

2. **For Independent Compatible Events:**
   - If events A and B are independent, the formula simplifies to:
     $$
     P(A \cap B) = P(A) \cdot P(B)
     $$
   - Since the occurrence of event A does not affect the probability of event B, $ P(B|A) = P(B) $.

3. **For Mutually Exclusive Events:**
   - If events A and B are mutually exclusive (or mutually exclusive events), then the probability of their intersection is always zero:
     $$
     P(A \cap B) = 0
     $$

### Examples

**Addition Theorem:**

- **Compatible Events:**
  - Event A: Rolling an even number on a die.
  - Event B: Rolling a number greater than 3 on a die.
  - The probability of rolling an even number or a number greater than 3:
    $$
    P(A \cup B) = P(A) + P(B) - P(A \cap B)
    $$
  - If $ P(A) = \frac{3}{6} = \frac{1}{2}$  and $P(B) = \frac{3}{6} = \frac{1}{2}$ , then  $P(A \cap B) = \frac{2}{6} = \frac{1}{3}$. Therefore:
    $$
    P(A \cup B) = \frac{1}{2} + \frac{1}{2} - \frac{1}{3} = \frac{2}{3}
    $$

- **Mutually Exclusive Events:**
  - Event A: Tossing a head on a coin.
  - Event B: Tossing a tail on a coin.
  - The probability of tossing a head or a tail:
   $$
    P(A \cup B) = P(A) + P(B)
   $$
  - Since these events cannot occur simultaneously, $ P(A \cap B) = 0 $ , and  
  $$
  P(A \cup B) = \frac{1}{2} + \frac{1}{2} = 1 
  $$.

**Multiplication Theorem:**

- **Compatible Dependent Events:**
  - Event A: Drawing an red Ace from a deck of cards (2 out of 52 cards).
  - Event B: Drawing a red card after drawing an red Ace (25 out of 51 remaining cards).
  - The probability of drawing an red Ace and then a red card:
    $$
    P(A \cap B) = P(A) \cdot P(B|A) = \frac{2}{52} \cdot \frac{25}{51}
    $$

- **Compatible Independent Events:**
  - Event A: Tossing a head on a coin.
  - Event B: Rolling an even number on a die.
  - The probability of tossing a head and rolling an even number:
    $$
    P(A \cap B) = P(A) \cdot P(B) = \frac{1}{2} \cdot \frac{1}{2} = \frac{1}{4}
    $$

- **Mutually Exclusive Events:**
  - Coin Toss: Let event A be tossing a head, and event B be tossing a tail. These events are mutually exclusive, as they cannot occur simultaneously:
    $$
    P(A \cap B) = 0
    $$

  - Die Roll: Event A is rolling a 2, and event B is rolling a 5. These events are mutually exclusive, as they cannot occur simultaneously:
    $$
    P(A \cap B) = 0
    $$

### Summary
- **Addition Theorem:**
  - For compatible events: $ P(A \cup B) = P(A) + P(B) - P(A \cap B) $
  - For mutually exclusive events: $ P(A \cup B) = P(A) + P(B) $

- **Multiplication Theorem:**
  - For depented compatible events: $ P(A \cap B) = P(A) \cdot P(B|A) $
  - For independent compatible events: $ P(A \cap B) = P(A) \cdot P(B) $
  - For mutually exclusive events: $ P(A \cap B) = 0 $


#### Python Implementation

```python
# Importing required library
from fractions import Fraction

# Define a function to compute the addition theorem for compatible events
def addition_compatible(P_A, P_B, P_A_and_B):
    return P_A + P_B - P_A_and_B

# Define a function to compute the addition theorem for mutually exclusive events
def addition_mutually_exclusive(P_A, P_B):
    return P_A + P_B

# Define a function to compute the multiplication theorem for compatible dependent events
def multiplication_dependent(P_A, P_B_given_A):
    return P_A * P_B_given_A

# Define a function to compute the multiplication theorem for independent events
def multiplication_independent(P_A, P_B):
    return P_A * P_B

# Define a function to compute the multiplication theorem for mutually exclusive events
def multiplication_mutually_exclusive():
    return 0

# Example probabilities
P_A = Fraction(1, 2)  # Probability of Event A
P_B = Fraction(1, 2)  # Probability of Event B
P_A_and_B = Fraction(1, 3)  # Probability of both A and B occurring (for compatible events)
P_B_given_A = Fraction(25, 51)  # Probability of B given A (for dependent events)

# Compute addition theorem for compatible events
result_addition_compatible = addition_compatible(P_A, P_B, P_A_and_B)

# Compute addition theorem for mutually exclusive events
result_addition_mutually_exclusive = addition_mutually_exclusive(P_A, P_B)

# Compute multiplication theorem for dependent events
result_multiplication_dependent = multiplication_dependent(P_A, P_B_given_A)

# Compute multiplication theorem for independent events
result_multiplication_independent = multiplication_independent(P_A, P_B)

# Compute multiplication theorem for mutually exclusive events
result_multiplication_mutually_exclusive = multiplication_mutually_exclusive()

# Print the results
print("Addition Theorem for Compatible Events: P(A ∪ B) =", result_addition_compatible)
print("Addition Theorem for Mutually Exclusive Events: P(A ∪ B) =", result_addition_mutually_exclusive)
print("Multiplication Theorem for Dependent Events: P(A ∩ B) =", result_multiplication_dependent)
print("Multiplication Theorem for Independent Events: P(A ∩ B) =", result_multiplication_independent)
print("Multiplication Theorem for Mutually Exclusive Events: P(A ∩ B) =", result_multiplication_mutually_exclusive)
```

### Mean vs Median

Both are measures of where the center of a data set lies, but they are usually different numbers. For example, take this list of numbers: 10,10,20,40,70.

- The ***mean*** (average) is found by adding all of the numbers together and dividing by the number of items in the set: 10 + 10 + 20 + 40 + 70 / 5 = 30.
    
- The ***median*** is found by ordering the set from lowest to highest and finding the exact middle. The median is just the middle number: 20.

Sometimes the two will be the same number. For example, the data set 1,2,4,6,7 has an average of 1 + 2 + 4 + 6 + 7 / 5 = 4 and a median (a middle) of 4.

```python
#Identify outliers using the IQR method: values outside the range [q1 - 1.5 * iqr, q3 + 1.5 * iqr].

def outlier(df, col):
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3-q1
    df_out = df[(df[col]>q3+1.5*iqr)|(df[col]<q1-1.5*iqr)]
    return df_out
```

Although the mean and median both give us some sense of the center of a distribution, ***they aren't always the same***. 

- The median always gives us a value that splits the data into two halves while the mean is a numeric average so extreme values can have a significant impact on the mean. 

    - In a symmetric distribution, the mean and median will be the same. 
    
Let's investigate with a density plot:

The probability density estimation (PDF) for a normal distribution is given by the formula:

$$
f(x) = \frac{1}{\sqrt{2 \pi} h} \sum_{i=1}^{n} \exp\left(-\frac{(x - x_i)^2}{2h^2}\right)
$$

Where:
- \( f(x) \) is the estimated density at point \( x \),
- \( n \) is the number of data points,
- \( x_i \) is each data point,
- \( h \) is the bandwidth parameter (a smoothing parameter).

```python
norm_data = pd.DataFrame(np.random.normal(size=100000))

norm_data.plot(kind="density", figsize=(10,10))
#norm_data.plot(kind="kde", figsize=(10,10))

# Plot black line at mean
plt.vlines(norm_data.mean(), ymin=0, ymax=0.4, linewidth=5.0, color = 'green')

# Plot red line at median
plt.vlines(norm_data.median(), ymin=0, ymax=0.4, linewidth=2.0, color="red")
```

In the plot the mean and median are both so close to zero that the red median line lies on top of the thicker black line drawn at the mean.

In skewed distributions, the mean tends to get pulled in the direction of the skew, while the median tends to resist the effects of skew:

```python
skewed_data = pd.DataFrame(np.random.exponential(size=100000))

skewed_data.plot(kind="density", figsize=(10,10), xlim=(-1,5))


plt.vlines(skewed_data.mean(), ymin=0, ymax=0.8, linewidth=5.0) # Plot black line at mean

plt.vlines(skewed_data.median(), ymin=0, ymax=0.8, linewidth=2.0, color="red") # Plot red line at median
```

The mean is also influenced heavily by outliers, while the median resists the influence of outliers:

```python
norm_data = np.random.normal(size=50)
outliers = np.random.normal(15, size=3)
combined_data = pd.DataFrame(np.concatenate((norm_data, outliers), axis=0))

combined_data.plot(kind="density", figsize=(10,10), xlim=(-5,20))


plt.vlines(combined_data.mean(), ymin=0, ymax=0.2, linewidth=5.0) # Plot black line at mean

plt.vlines(combined_data.median(), ymin=0, ymax=0.2, linewidth=2.0, color="red")   # Plot red line at median
```

Since the median tends to resist the effects of skewness and outliers, it is known a "robust" statistic. The median generally gives a better sense of the typical value in a distribution with significant skew or outliers.

### Variance and Standard Deviation

Variance and standard deviation are two other common ***measures of spread.*** 


The **variance** of a distribution(population) is the ***average of the squared deviations (differences) from the mean(expectation value)***. 

$$variance = \sigma^2 = \frac{\displaystyle\sum_{i=1}^{n}(x_i - \bar{x})^2} {n}$$

<!---![alt text](https://wikimedia.org/api/rest_v1/media/math/render/svg/00eb0cde84f0a838a2de6db9f382866427aeb3bf)-->

Use `df.var()` to check variance of sample data that is a subset of the entire population:

```python
#To check variance of sample data that is a subset of the entire population
def var(df, col):
    mean = df[col].mean()
    n = len(df)
    l = np.sum([(x-mean)**2/(n-1) for x in df[col]])
    return l
```

### Statistical Measures in Data Analysis

### Skewness and Kurtosis

Beyond measures of center and spread, descriptive statistics include measures that give you a sense of the shape of a distribution.

- [**Skewness**](https://en.wikipedia.org/wiki/Skewness) measures the skew or asymmetry of a distribution.
- [**Kurtosis**](https://en.wikipedia.org/wiki/Kurtosis) measures the "peakedness" of a distribution.  

#### Variance

Variance measures the average squared deviation of each data point from the mean. It is calculated using the formula:
$$
 \sigma^2 = \frac{\sum_{i=1}^{n} (x_i - \bar{x})^2}{n}
$$

#### Skewness

Skewness measures the asymmetry or lack of symmetry in a distribution. It involves cubing the deviations from the mean before averaging. The formula for skewness is:

$$ S = \frac{\sum_{i=1}^{n} (x_i - \bar{x})^3}{n \cdot \sigma^3} $$

Positive skewness indicates a longer right tail, while negative skewness indicates a longer left tail.

The **absolute value of skewness** provides information about the degree of asymmetry:

- **|Skewness| < 0.5**: The distribution is approximately symmetric. There's little skew, and the mean and median are close.

- **0.5 < |Skewness| < 1**: The distribution is moderately skewed. The skew is noticeable but not extreme.

- **|Skewness| > 1**: The distribution is highly skewed. The skew is strong, meaning the data is significantly pulled in the direction of the skew.

- **|Skewness| = 0**: The distribution is not skewed. So the distribution of data is Normal and, mean and median is equal. 


#### Kurtosis

Kurtosis measures the "tailedness" or shape of the distribution. It involves raising the deviations from the mean to the 4th power before averaging, and subtracting 3 as a normalization step. The formula for kurtosis is:

$$ K = \frac{\sum_{i=1}^{n} (x_i - \bar{x})^4}{n \cdot \sigma^4} - 3$$

Positive kurtosis indicates heavy tails and a more peaked distribution, while negative kurtosis indicates light tails and a flatter distribution.


- **Kurtosis > 0 (Leptokurtic)**: Indicates a distribution with heavier tails and a sharper peak compared to a normal distribution. Extreme values are more frequent.
  
- **Kurtosis < 0 (Platykurtic)**: Indicates a distribution with lighter tails and a flatter peak compared to a normal distribution. Extreme values are less frequent.

- **Kurtosis = 0 (Mesokurtic)**: Indicates a distribution similar to a normal distribution, with tails and peak similar to the standard bell curve.

##### Absolute Value of Kurtosis

The absolute value of kurtosis can provide insight into the deviation from normality:

- **|Kurtosis| < 1**: The distribution's tails are not significantly different from a normal distribution. It is approximately mesokurtic.
  
- **1 ≤ |Kurtosis| < 3**: The distribution shows some degree of deviation, indicating either slightly heavier or lighter tails. The distribution is slightly leptokurtic or platykurtic.
  
- **|Kurtosis| ≥ 3**: The distribution shows a significant deviation from normality. It is either strongly leptokurtic (heavy tails) or strongly platykurtic (light tails).

To explore these two measures further, let's create some dummy data and inspect it:

```python
norm_data = np.random.normal(size=100000)
skewed_data = np.concatenate((np.random.normal(size=35000)+2, 
                             np.random.exponential(size=65000)), 
                             axis=0)
uniform_data = np.random.uniform(1,2, size=100000)
peaked_data = np.concatenate((np.random.exponential(size=50000),
                             np.random.exponential(size=50000)*(-1)),
                             axis=0)

data_df = pd.DataFrame({"norm":norm_data,
                       "skewed":skewed_data,
                       "uniform":uniform_data,
                       "peaked":peaked_data})
```

```python
data_df.plot(kind="density",
            figsize=(10,10),
            xlim=(-5,5))
```

Now let's check the skewness of each of the distributions. Since skewness measures asymmetry, we'd expect to see low skewness for all of the distributions except the skewed one, because all the others are roughly symmetric:

```python
data_df.skew()
```

Now let's check kurtosis. Since kurtosis measures peakedness, we'd expect the flat (uniform) distribution have low kurtosis while the distributions with sharper peaks should have higher kurtosis.

```python
data_df.kurt()
```

As we can see from the output, 
- The normally distributed data has a kurtosis near zero
- The flat distribution has negative kurtosis and 
- The two pointier distributions have positive kurtosis.
