# The Wilcoxon Test

**`Wilcoxon Tests`** are non-parametric statistical tests used to compare differences between groups or measurements when the assumptions of parametric tests (like the t-test) are not met. These tests are useful for data that does not follow a normal distribution or is ordinal in nature.

### Types of Wilcoxon Tests

**1. Wilcoxon Rank-Sum Test (Mann-Whitney U Test)**
- **Purpose**: Compares differences between two independent samples.

**2. Wilcoxon Signed-Rank Test**
- **Purpose**: Used to test the differences between two related or paired samples.

### Key Points

- **Non-Parametric**: Does not require the assumption of normality in data.
- **Ranks**: Both tests use ranking of data rather than actual values to determine differences.

## **The Wilcoxon Rank Sum Test(The Mann-Whitney U Test)** 
   This is often described as the non-parametric version of the two-sample t-test. You sometimes see it in analysis flowcharts after a question such as "is your data normal?" A "no" branch off this question will recommend a Wilcoxon test if you're comparing two groups of continuous measures.

The Wilcoxon Rank Sum Test is a common statistical test that is used in many fields including economics, biological sciences and epidemiology. It is particularly useful when you are assessing the difference between two independent groups with low numbers of individuals in each group (usually less than 30), which are not normally distributed, and where the data are continuous. If you are interested in comparing more than two groups which have skewed data, a Kruskal-Wallis One-Way analysis of variance (ANOVA) should be used.

### The Wilcoxon Rank Sum Test Assumptions

Some key assumptions for the Mann-Whitney U Test are detailed below:

1. **Continuous Variable**: 
   - The variable being compared between the two groups must be continuous (able to take any number in a range – for example age, weight, height, or heart rate). This is because the test is based on ranking the observations in each group.

2. **Non-Normal Distribution**: 
   - The data are assumed to take a non-Normal, or skewed, distribution. If your data are normally distributed, Two-Sample Student’s t-test should be used to compare the two groups instead.

3. **Similar Shape Across Groups**: 
   - While the data in both groups are not assumed to be Normal, the data are assumed to be similar in shape across the two groups.

4. **Independent Samples**: 
   - The data should be two randomly selected independent samples, meaning the groups have no relationship to each other. If samples are paired (for example, two measurements from the same group of participants), then a paired samples t-test should be used instead.

5. **Sufficient Sample Size**: 
   - A valid test requires a sufficient sample size, usually more than 5 observations in each group.

### 2. When to Use It

- **Non-Normal Data**: If your data doesn't follow a normal distribution, the two-sample t-test may not be appropriate because it relies on that assumption. In such cases, you can use the Wilcoxon Rank Sum Test.
- **Ordinal Data**: It can also be used if your data is ordinal (i.e., data that can be ranked but not measured on a continuous scale).

### 3. How It Works

- **Ranking the Data**: Instead of comparing the actual values of the data, the Wilcoxon test ranks all the values from both groups together.
- **Sum of Ranks**: It then calculates the sum of the ranks for each group.
- **Comparison**: The test compares these rank sums to determine if the groups are significantly different.

### The Wilcoxon Rank Sum Test Example

Consider a randomized controlled trial evaluating a new anti-retroviral therapy for HIV. A pilot trial randomly assigned participants to either the treated or untreated groups (N=14). We want to assess the viral load (quantity of virus per milliliter of blood) in the treated versus the untreated groups. In practice, a Mann-Whitney U Test would be easily and quickly calculated using statistical software such as SPSS or Stata, but the steps are laid out below.

The first step is to assign ranks to the values from the full sample (both treatment groups pooled together) in order from smallest to largest. We can then generate a test statistic based on the ranks.

#### The Wilcoxon Rank Sum Test Statistic
After summing the ranks for each group, The Wilcoxon Rank Sum test statistic is selected as the smallest of the two following calculated U values:

##### Symbols:
- **$ n_1 $**: Number of observations in Group 1.
- **$ n_2 $**: Number of observations in Group 2.
- **$ R_1 $**: Sum of ranks for Group 1.
- **$ R_2 $**: Sum of ranks for Group 2.

##### Formula for $ U_1 $:
$$
U_1 = n_1 \times n_2 + \frac{n_1 \times (n_1 + 1)}{2} - R_1
$$

**Step-by-Step:**
1. Multiply the number of observations in Group 1 ($ n_1$) by the number of observations in Group 2 ($ n_2 $).
2. Add the product of $ \frac{n_1 \times (n_1 + 1)}{2} $, which represents the sum of ranks for Group 1 if all its ranks were perfectly sequential (like 1, 2, 3, etc.).
3. Subtract the actual sum of ranks $ R_1 $ for Group 1.

#### Formula for $ U_2 $:
$$
U_2 = n_1 \times n_2 + \frac{n_2 \times (n_2 + 1)}{2} - R_2
$$

**Step-by-Step:**
1. Multiply the number of observations in Group 1 ($ n_1 $) by the number of observations in Group 2 ($ n_2 $).
2. Add the product of $ \frac{n_2 \times (n_2 + 1)}{2} $, which represents the sum of ranks for Group 2 if all its ranks were perfectly sequential.
3. Subtract the actual sum of ranks $ R_2 $ for Group 2.

#### What the Formulas Do

1. **Perfectly Sequential Sum:**
   - The term $\frac{n_1 \times (n_1 + 1)}{2}$ represents what the sum of ranks for Group 1 would be if all its ranks were in a perfectly sequential order (like 1, 2, 3, ...).

2. **Actual Rank Sum:**
   - The values $R_1$ and $R_2$ are the actual sums of ranks for the two groups. These sums are based on the mixed ranks after combining both groups.

3. **U Value Calculation:**
   - The formulas calculate how far the actual rank sums are from what they would be in a perfectly sequential scenario.

4. **Choosing the Smallest U:**
   - The test uses the smallest($min(U_1,U_2)$) U value to determine if there is a significant difference between the two groups.

#### Summary

- **Perfectly Sequential:** Is the ideal scenario where ranks are consecutive integers.
- **Actual Ranks:** Reflect the real data, which is usually mixed.
- **U Values:** Compare actual rank sums to this ideal scenario to test for differences.


Where we let 1 denote the treated group and 2 denote the untreated group (denotation of groups is arbitrary), where $ n_1 $ and $ n_2 $ are the number of participants, and where $ R_1 $ and $ R_2 $ are the sums of the ranks in the treated and untreated groups, respectively.

In this example:

- $ U_1 = 40 $
- $ U_2 = 9 $

We therefore select **U = 9** as the test statistic.


#### Critical Value of U

To determine the critical value of $ U $ for the Mann-Whitney U Test, we use a reference table of critical values. You can find such a table at the following link:

##### Example of How to Use the Table:

1. **Determine Sample Sizes:**
   - For this example, the sample sizes are $ n_1 = 7 $ and $ n_2 = 7 $.

2. **Select Significance Level:**
   - The significance level (α) is $ 0.05 $ for a two-tailed test.

3. **Locate the Critical Value:**
   - Find the row for $ n_1 = 7 $ and the column for $ n_2 = 7 $ in the table.
   - Look up the critical value corresponding to the chosen significance level.

4. **Compare Test Statistic:**
   - Compare your calculated U value with the critical value from the table.
   - If the calculated U is less than or equal to the critical value, you reject the null hypothesis.

```python
from scipy.stats import mannwhitneyu

# Data
group_a = [540, 670, 1000, 960, 1200, 4650, 4200]
group_b = [5000, 4300, 1300, 900, 7400, 4500, 7500]

# Perform Mann-Whitney U Test
U, p_value = mannwhitneyu(group_a, group_b, alternative='two-sided')

print(f"Test Statistic (U): {U:.2f}")
print(f"P-Value: {p_value:.4f}")
```

## Wilcoxon Signed-Rank Test

The Wilcoxon Signed-Rank Test is a non-parametric statistical test used to compare two related samples or paired observations. It is often used when the data does not meet the assumptions of the paired t-test, particularly the assumption of normality.

### Key Points of the Wilcoxon Signed-Rank Test

#### 1. Purpose
- The test is used to determine if there is a significant difference between two related groups. For example, it might be used to assess whether a treatment has an effect by comparing measurements taken before and after the treatment on the same subjects.

#### 2. When to Use It
- **Paired Data**: The test is suitable when you have two sets of related or paired data, such as pre-treatment and post-treatment measurements.
- **Non-Normal Data**: It is used when the differences between paired observations are not normally distributed.

#### 3. Steps in the Wilcoxon Signed-Rank Test

- **Calculate Differences**:
  Compute the difference between each pair of observations (e.g., post-treatment value - pre-treatment value).

- **Rank Absolute Differences**:
  Rank the absolute values of these differences from smallest to largest. The ranks are assigned in ascending order, with tied differences receiving the average rank of their positions.

- **Assign Signs**:
  Reassign the original signs (positive or negative) to the ranks based on the sign of the differences.

- **Sum Ranks**:
  Compute the sum of the ranks for the positive differences and the sum for the negative differences.

- **Test Statistic**:
  The test statistic is the smaller of these two sums. This statistic is compared against a critical value from the Wilcoxon distribution to determine significance.

#### 4. Hypothesis
- **Null Hypothesis (H₀)**: The median of the differences between the paired observations is zero (no effect).
- **Alternative Hypothesis (H₁)**: The median of the differences is not zero (there is an effect).

#### 5. Example Calculation

Suppose you have the following paired data before and after treatment:
- Before: [5.0, 6.0, 7.4, 8.4]
- After: [5.5, 6.8, 7.1, 8.2]

1. **Calculate Differences**:
   - Differences: [0.5, 0.8, -0.3, -0.2]

2. **Rank Absolute Differences**:
   - Absolute Differences: [0.5, 0.8, 0.3, 0.2]
   - Ranks: [3, 4, 2, 1]

3. **Assign Signs and Sum Ranks**:
   - Signed Ranks: [3, 4, -2, -1]
   - Sum of Positive Ranks: 3 + 4 = 7
   - Sum of Negative Ranks: -1 - 2 = -3

4. **Test Statistic**:
   - The test statistic is the smaller of the sum of positive ranks and the absolute value of the sum of negative ranks. In this case, it would be 3.

```python
# Software Implementation
"""
In Python, you can use `scipy.stats.wilcoxon()` to perform the test. Here’s an example code snippet:
"""

from scipy.stats import wilcoxon

# Paired data
before = [5.0, 6.0, 7.4, 8.4]
after = [5.5, 6.8, 7.1, 8.2]

# Perform the Wilcoxon Signed-Rank Test
stat, p_value = wilcoxon(before, after)

print(f"Test Statistic: {stat:.2f}")
print(f"P-Value: {p_value:.4f}")
```
