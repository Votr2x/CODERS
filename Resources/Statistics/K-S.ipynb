```python
# ECDF
```

```python
data = np.random.normal(loc=0, scale=1, size=100)

x = np.sort(data)
y = np.arange(1, len(x)+1) / len(x)

plt.figure(figsize=(6,4))
plt.plot(x, y, marker=".", linestyle="none")
plt.xlabel("Value")
plt.ylabel("ECDF")
plt.title("Empirical Cumulative Distribution Function (ECDF)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

my_numbers = np.random.normal(0, 1, 100)  
my_guess = np.random.normal(0, 1, 100)    

ks_stat, p_value = stats.kstest(my_numbers, 'norm')
print("Difference score:", ks_stat)
print("Chance your guess is right:", p_value)

ks_stat2, p_value2 = stats.ks_2samp(my_numbers, my_guess)
print("Difference score:", ks_stat2)
print("Chance the two bags are same:", p_value2)

def plot_line(numbers, name):
    sorted_nums = np.sort(numbers)
    yvals = np.arange(1, len(sorted_nums)+1)/len(sorted_nums)
    plt.plot(sorted_nums, yvals, label=name)

plot_line(my_numbers, "My Bag")
plot_line(my_guess, "Guess Bag")
plt.legend()
plt.show()
```
