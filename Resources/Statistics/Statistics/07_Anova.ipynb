# Analysis of Variance (ANOVA)

## What is an Analysis of Variance?

An **Analysis of Variance (ANOVA)** tests whether statistically significant differences exist between more than two samples. For this purpose, the means and variances of the respective groups are compared with each other. 

In contrast to the **t-test**, which tests whether there is a difference between two samples, the ANOVA tests whether there is a difference between more than two groups.

### Why not calculate multiple t-tests?
ANOVA is used when there are more than two groups. Of course, it would also be a possibility to calculate a t-test for each combination of the groups. The problem here, however, is that every hypothesis test has some degree of error. This probability of error is usually set at 5%, so that, from a purely statistical point of view, every 20th test gives a wrong result

If, for example, 20 groups are compared in which there is actually no difference, one of the tests will show a significant difference purely due to the sampling.

## Hypothesis Testing in ANOVA

**Analysis of Variance (ANOVA)** is a statistical method used to determine if there are any statistically significant differences between the means of three or more independent groups. ANOVA relies on hypothesis testing to make this determination. Below, we’ll explore the different types of ANOVA through the lens of hypothesis testing.

- **One-factor (or one-way) ANOVA**
- **Two-factors (or two-way) ANOVA**
- **One-Way ANOVA with Repeated Measures**
- **Two-Way ANOVA with Repeated Measures**

### Difference between one-way and two-way ANOVA:
The one-way analysis of variance only checks whether an independent variable has an influence on a metric dependent variable. This is the case, for example, if it is to be examined whether the place of residence (independent variable) has an influence on the salary (dependent variable). However, if two factors, i.e. two independent variables, are considered, a two-way analysis of variance must be used.

### 1. One-factor (or one-way) ANOVA

**One-Way ANOVA** is used when you want to test if there are significant differences between the means of three or more groups based on one independent variable.

- **Null Hypothesis ($H_0$)**: The means of all groups are equal.
  $$
  H_0: \mu_1 = \mu_2 = \mu_3 = \ldots = \mu_k
  $$
  
- **Alternative Hypothesis ($H_a$)**: At least one group mean is different.
  $$
  H_a: \text{At least one } \mu_i \text{ is different.}
  $$

The **one-way ANOVA** tests these hypotheses by comparing the variance within each group to the variance between the groups.

#### Understanding Variance in One-Way ANOVA

**Variance within each group**: This refers to how much the individual data points within a single group differ from the group mean. For example, if you have a group of students who took the same test, the variance within this group would measure how different each student's score is from the average score of that group.

**Variance between the groups**: This refers to the differences in the means (averages) of the different groups. For example, if you have three different classes of students, each taught using a different teaching method, the variance between the groups would measure how different the average scores are across these three classes.

#### The Logic of One-Way ANOVA

The one-way ANOVA works by comparing these two types of variance:

- **Variance Within Groups**: This is sometimes called the "error variance" or "residual variance." It reflects the variation in scores that cannot be explained by the factor you're testing (e.g., different teaching methods). If the scores within each group are very spread out, this variance will be large.

- **Variance Between Groups**: This is sometimes called the "explained variance" or "treatment variance." It reflects the variation in group means, which is due to the factor you're testing (e.g., different teaching methods). If the groups have very different averages, this variance will be large.

#### The F-Ratio

ANOVA calculates an F-ratio, which is the ratio of the variance between the groups to the variance within the groups:
$$
F = \frac{\text{Variance between groups}}{\text{Variance within groups}}
$$

- If the variance between the groups is much larger than the variance within the groups, the F-ratio will be large, suggesting that the group means are different enough to be statistically significant.
- If the F-ratio is close to 1, it suggests that the variance between groups is not much greater than the variance within groups, meaning any observed differences in group means might just be due to random variation rather than a real effect.


##### Example: ANOVA Calculation

Suppose you are comparing the effectiveness of three different diets on weight loss:

- **Groups**: Diet A, Diet B, Diet C.
- **Hypotheses**:
  - $ H_0 $: The mean weight loss is the same for all diets.
  - $ H_1 $: At least one diet leads to a different mean weight loss.

Let's assume we have the following weight loss data (in pounds) for each diet:

- **Diet A**: [5, 7, 6, 8, 7]
- **Diet B**: [8, 9, 6, 10, 9]
- **Diet C**: [6, 7, 5, 6, 7]

We will calculate the F-ratio to determine if there is a significant difference between the means.

```python
# Step 1: Calculate Group Means
import numpy as np

diet_a = np.array([5, 7, 6, 8, 7])
diet_b = np.array([8, 9, 6, 10, 9])
diet_c = np.array([6, 7, 5, 6, 7])

mean_a = np.mean(diet_a)
mean_b = np.mean(diet_b)
mean_c = np.mean(diet_c)

mean_a, mean_b, mean_c
```

```python
#Step 2: Calculate the Overall Mean

overall_mean = np.mean(np.concatenate([diet_a, diet_b, diet_c]))
overall_mean
```

```python
#Step 3: Calculate Variance Between Groups(Sum of Squares Between Groups)

n_a = len(diet_a)
n_b = len(diet_b)
n_c = len(diet_c)

ss_between = n_a * (mean_a - overall_mean)**2 + n_b * (mean_b - overall_mean)**2 + n_c * (mean_c - overall_mean)**2
ss_between
```

```python
#Step 4: Calculate Variance Within Groups(Sum of Squares Within Groups )

ss_within = np.sum((diet_a - mean_a)**2) + np.sum((diet_b - mean_b)**2) + np.sum((diet_c - mean_c)**2)
ss_within
```

**Degrees of Freedom (df):**
- **df_between:** Degrees of freedom between groups, calculated as the number of groups minus 1.
  - Formula: df_between = k - 1, where k is the number of groups.
- **df_within:** Degrees of freedom within groups, calculated as the total number of observations minus the number of groups.
  - Formula: df_within = N - k, where N is the total number of observations and k is the number of groups.

```python
#Step 5: Calculate the F-Ratio

df_between = 2  # Number of groups - 1
df_within = len(diet_a) + len(diet_b) + len(diet_c) - 3  # Total observations - Number of groups

ms_between = ss_between / df_between
ms_within = ss_within / df_within

f_ratio = ms_between / ms_within
f_ratio
```

To determine whether to reject or fail to reject the null hypothesis, we compare the calculated F-ratio to the critical value of F from the F-distribution table at a chosen significance level (e.g., 0.05) with the appropriate degrees of freedom.

- **If the F-ratio is greater than the critical value,** we reject the null hypothesis, indicating that there is a statistically significant difference in weight loss between at least one pair of diets.
- **If the F-ratio is less than or equal to the critical value,** we fail to reject the null hypothesis, indicating that any differences in weight loss are likely due to random variation.

```python
from scipy.stats import f

alpha = 0.05
f_critical = f.ppf(1 - alpha, df_between, df_within)
f_critical
```

#### Understanding the P-Value in the Context of ANOVA

The **p-value** is a critical part of hypothesis testing in ANOVA. It tells you the probability of observing your data, or something more extreme, if the null hypothesis (\( H_0 \)) is true.

- **Small p-value (typically ≤ 0.05)**: Indicates strong evidence against the null hypothesis, so you reject the null hypothesis.
- **Large p-value (> 0.05)**: Indicates weak evidence against the null hypothesis, so you fail to reject the null hypothesis.

```python
# Calculate the p-value based on the F-ratio
p_value = 1 - f.cdf(f_ratio, df_between, df_within)
p_value
```

If you have the data and just want to run a one-way ANOVA, you can use scipy.stats.f_oneway to get the F-ratio and p-value:

```python
from scipy import stats

#f_statistic(f_ratio)
f_statistic, p_value = stats.f_oneway(diet_a, diet_b, diet_c)

print(f"F-statistic: {f_statistic}")
print(f"P-value: {p_value}")
```

### 2. Two-factors (or two-way) ANOVA

**Two-Way ANOVA** is used to test the effect of two independent variables on a dependent variable, as well as the interaction between the two variables.

- **Null Hypotheses**:
  - $H_0^1$: The means are equal across the levels of the first factor.
  - $H_0^2$: The means are equal across the levels of the second factor.
  - $H_0^3$: There is no interaction between the two factors.

- **Alternative Hypotheses**:
  - $H_a^1$: The means differ across the levels of the first factor.
  - $H_a^2$: The means differ across the levels of the second factor.
  - $H_a^3$: There is an interaction between the two factors.

The **two-way ANOVA** tests these hypotheses by analyzing the main effects of each factor and their interaction.

**Example:**

You are studying the impact of two factors—exercise type (running, swimming, cycling) and diet type (low-carb, low-fat)—on weight loss.

#### Key Concepts

##### Main Effects

- **Effect of Factor 1 (e.g., Exercise Type)**: Tests whether the mean weight loss differs significantly across the levels of exercise type (e.g., Running, Swimming, Cycling).

- **Effect of Factor 2 (e.g., Diet Type)**: Tests whether the mean weight loss differs significantly across the levels of diet type (e.g., Low-Carb, Low-Fat).

##### Interaction Effect

- **Interaction Effect**: Tests whether the effect of one factor (e.g., Exercise Type) on the dependent variable (e.g., Weight Loss) depends on the level of the other factor (e.g., Diet Type).

#### Hypotheses

##### For Exercise Type (Factor 1)

- **Null Hypothesis ($H_0^1$)**: The mean weight loss is the same for all exercise types.

- **Alternative Hypothesis ($H_a^1$)**: At least one exercise type leads to a different mean weight loss.

##### For Diet Type (Factor 2)

- **Null Hypothesis ($H_0^2$)**: The mean weight loss is the same for all diet types.

- **Alternative Hypothesis ($H_a^2$)**: At least one diet type leads to a different mean weight loss.

##### For Interaction

- **Null Hypothesis ($H_0^3$)**: There is no interaction between exercise type and diet type (the effect of exercise type on weight loss is the same across diet types).

- **Alternative Hypothesis ($H_a^3$)**: There is an interaction between exercise type and diet type (the effect of exercise type on weight loss depends on diet type).

#### Calculation Example

##### Scenario
You have data on weight loss for three exercise types and two diet types. The data is organized as follows:

| Exercise Type | Diet Type | Weight Loss (kg) |
|---------------|-----------|------------------|
| Running       | Low-Carb  | 5.2              |
| Running       | Low-Carb  | 4.8              |
| Running       | Low-Fat   | 6.1              |
| Running       | Low-Fat   | 5.9              |
| Swimming      | Low-Carb  | 4.5              |
| Swimming      | Low-Carb  | 4.7              |
| Swimming      | Low-Fat   | 6.3              |
| Swimming      | Low-Fat   | 6.0              |
| Cycling       | Low-Carb  | 5.0              |
| Cycling       | Low-Carb  | 5.2              |
| Cycling       | Low-Fat   | 6.0              |
| Cycling       | Low-Fat   | 5.8              |

#### Step-by-Step Calculation

##### 1. Calculate Means

- **Overall Mean**: Average weight loss across all groups.
  
  $$ \text{Overall Mean} = \frac{\sum \text{Weight Loss}}{\text{Total Number of Observations}} $$

- **Group Means**:
  - **Exercise Type Means**: Mean weight loss for each exercise type (e.g., Running, Swimming, Cycling).
    
    $$ \text{Mean}_{\text{Exercise}} = \frac{\sum \text{Weight Loss (for each Exercise Type)}}{\text{Number of Observations for each Exercise Type}} $$
  
  - **Diet Type Means**: Mean weight loss for each diet type (e.g., Low-Carb, Low-Fat).
    
    $$ \text{Mean}_{\text{Diet}} = \frac{\sum \text{Weight Loss (for each Diet Type)}}{\text{Number of Observations for each Diet Type}} $$
  
  - **Interaction Means**: Mean weight loss for each combination of exercise type and diet type.
    
    $$ \text{Mean}_{\text{Interaction}} = \frac{\sum \text{Weight Loss (for each Exercise Type and Diet Type Combination)}}{\text{Number of Observations for each Combination}} $$

##### 2. Calculate Sum of Squares

- **SS_between_exercise**: Variability due to exercise type.
  
  $$ \text{SS}_{\text{between\_exercise}} = \sum_{i} n_{i} \times (\text{Mean}_{i} - \text{Overall Mean})^2 $$

- **SS_between_diet**: Variability due to diet type.
  
  $$ \text{SS}_{\text{between\_diet}} = \sum_{j} n_{j} \times (\text{Mean}_{j} - \text{Overall Mean})^2 $$

- **SS_interaction**: Variability due to interaction between exercise type and diet type.
  
  $$ \text{SS}_{\text{interaction}} = \sum_{i,j} n_{i,j} \times (\text{Mean}_{i,j} - \text{Mean}_{i} - \text{Mean}_{j} + \text{Overall Mean})^2 $$

- **SS_within**: Variability within groups (error).
  
  $$ \text{SS}_{\text{within}} = \sum_{i} \sum_{j} \sum_{k} (\text{Weight Loss}_{ijk} - \text{Mean}_{ij})^2 $$

##### 3. Calculate Degrees of Freedom

- **df_between_exercise**: Number of exercise types - 1.
  
  $$ \text{df}_{\text{between\_exercise}} = \text{Number of Exercise Types} - 1 $$

- **df_between_diet**: Number of diet types - 1.
  
  $$ \text{df}_{\text{between\_diet}} = \text{Number of Diet Types} - 1 $$

- **df_interaction**: (Number of exercise types - 1) × (Number of diet types - 1).
  
  $$ \text{df}_{\text{interaction}} = (\text{Number of Exercise Types} - 1) \times (\text{Number of Diet Types} - 1) $$

- **df_within**: Total number of observations - (Number of exercise types × Number of diet types).
  
  $$ \text{df}_{\text{within}} = \text{Total Number of Observations} - (\text{Number of Exercise Types} \times \text{Number of Diet Types}) $$

##### 4. Calculate Mean Squares

- **MS_between_exercise**: 

  $$ \text{MS}_{\text{between\_exercise}} = \frac{\text{SS}_{\text{between\_exercise}}}{\text{df}_{\text{between\_exercise}}} $$

- **MS_between_diet**: 

  $$ \text{MS}_{\text{between\_diet}} = \frac{\text{SS}_{\text{between\_diet}}}{\text{df}_{\text{between\_diet}}} $$

- **MS_interaction**: 

  $$ \text{MS}_{\text{interaction}} = \frac{\text{SS}_{\text{interaction}}}{\text{df}_{\text{interaction}}} $$

- **MS_within**: 

  $$ \text{MS}_{\text{within}} = \frac{\text{SS}_{\text{within}}}{\text{df}_{\text{within}}} $$

##### 5. Calculate F-Ratios

- **F_exercise**: 

  $$ \text{F}_{\text{exercise}} = \frac{\text{MS}_{\text{between\_exercise}}}{\text{MS}_{\text{within}}} $$

- **F_diet**: 

  $$ \text{F}_{\text{diet}} = \frac{\text{MS}_{\text{between\_diet}}}{\text{MS}_{\text{within}}} $$

- **F_interaction**: 

  $$ \text{F}_{\text{interaction}} = \frac{\text{MS}_{\text{interaction}}}{\text{MS}_{\text{within}}} $$

##### 6. Interpret Results

   - Compare the calculated F-ratio to the critical value from the F-distribution table to determine if the result is statistically significant.
   - Use the p-value to assess the strength of evidence against the null hypotheses.


```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# Sample data
data = {
    'Exercise Type': ['Running', 'Running', 'Running', 'Running',
                      'Swimming', 'Swimming', 'Swimming', 'Swimming',
                      'Cycling', 'Cycling', 'Cycling', 'Cycling'],
    'Diet Type': ['Low-Carb', 'Low-Carb', 'Low-Fat', 'Low-Fat',
                  'Low-Carb', 'Low-Carb', 'Low-Fat', 'Low-Fat',
                  'Low-Carb', 'Low-Carb', 'Low-Fat', 'Low-Fat'],
    'Weight Loss': [5.2, 4.8, 6.1, 5.9,
                    4.5, 4.7, 6.3, 6.0,
                    5.0, 5.2, 6.0, 5.8]
}

df = pd.DataFrame(data)

# Calculate overall mean
overall_mean = df['Weight Loss'].mean()

# Calculate means for each combination of Exercise Type and Diet Type
means = df.groupby(['Exercise Type', 'Diet Type'])['Weight Loss'].mean().unstack()

# Number of observations per combination
n = df.groupby(['Exercise Type', 'Diet Type']).size().unstack().fillna(0)

# Calculate means for each factor
exercise_means = df.groupby('Exercise Type')['Weight Loss'].mean()
diet_means = df.groupby('Diet Type')['Weight Loss'].mean()

# Sum of Squares Between Groups (SSB) for Exercise Type
ss_between_exercise = sum(n.loc[exercise].sum() * (exercise_means[exercise] - overall_mean)**2
                          for exercise in exercise_means.index)

# Sum of Squares Between Groups (SSB) for Diet Type
ss_between_diet = sum(n[diet].sum() * (diet_means[diet] - overall_mean)**2
                      for diet in diet_means.index)

# Sum of Squares Between Groups (SSB) for Interaction
ss_between_interaction = sum(
    n.loc[exercise, diet] * (means.loc[exercise, diet] - exercise_means[exercise] - diet_means[diet] + overall_mean)**2
    for exercise in exercise_means.index
    for diet in diet_means.index
)

# Calculate Sum of Squares Within Groups (SSW)
ss_within = sum(df[df['Exercise Type'] == ex].groupby('Diet Type')['Weight Loss'].apply(lambda x: np.sum((x - means.loc[ex, x.name])**2)).sum()
                for ex in exercise_means.index) + \
            sum(df[df['Diet Type'] == di].groupby('Exercise Type')['Weight Loss'].apply(lambda x: np.sum((x - means.loc[x.name, di])**2)).sum()
                for di in diet_means.index)

# Degrees of Freedom
df_between_exercise = len(exercise_means) - 1
df_between_diet = len(diet_means) - 1
df_interaction = df_between_exercise * df_between_diet
df_within = len(df) - (len(exercise_means) * len(diet_means))

# Mean Squares
ms_between_exercise = ss_between_exercise / df_between_exercise
ms_between_diet = ss_between_diet / df_between_diet
ms_interaction = ss_between_interaction / df_interaction
ms_within = ss_within / df_within

# F-Ratios
f_ratio_exercise = ms_between_exercise / ms_within
f_ratio_diet = ms_between_diet / ms_within
f_ratio_interaction = ms_interaction / ms_within

# P-values
p_value_exercise = stats.f.sf(f_ratio_exercise, df_between_exercise, df_within)
p_value_diet = stats.f.sf(f_ratio_diet, df_between_diet, df_within)
p_value_interaction = stats.f.sf(f_ratio_interaction, df_interaction, df_within)

# Output
print(f"Sum of Squares Between Exercise Types: {ss_between_exercise:.2f}")
print(f"Sum of Squares Between Diet Types: {ss_between_diet:.2f}")
print(f"Sum of Squares Between Interaction: {ss_between_interaction:.2f}")
print(f"Sum of Squares Within Groups: {ss_within:.2f}")
print(f"Mean Square Between Exercise Types: {ms_between_exercise:.2f}")
print(f"Mean Square Between Diet Types: {ms_between_diet:.2f}")
print(f"Mean Square Between Interaction: {ms_interaction:.2f}")
print(f"Mean Square Within Groups: {ms_within:.2f}")
print(f"F-Ratio for Exercise Type: {f_ratio_exercise:.2f}")
print(f"F-Ratio for Diet Type: {f_ratio_diet:.2f}")
print(f"F-Ratio for Interaction: {f_ratio_interaction:.2f}")
print(f"P-Value for Exercise Type: {p_value_exercise:.4f}")
print(f"P-Value for Diet Type: {p_value_diet:.4f}")
print(f"P-Value for Interaction: {p_value_interaction:.4f}")

# Interpretation
alpha = 0.05
print("\nInterpretation:")
if p_value_exercise < alpha:
    print("Reject the null hypothesis for Exercise Type.")
else:
    print("Fail to reject the null hypothesis for Exercise Type.")

if p_value_diet < alpha:
    print("Reject the null hypothesis for Diet Type.")
else:
    print("Fail to reject the null hypothesis for Diet Type.")

if p_value_interaction < alpha:
    print("Reject the null hypothesis for Interaction.")
else:
    print("Fail to reject the null hypothesis for Interaction.")
```

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Create a DataFrame with the sample data
data = {
    'Exercise': ['Running', 'Running', 'Running', 'Running', 'Swimming', 'Swimming', 'Swimming', 'Swimming', 'Cycling', 'Cycling', 'Cycling', 'Cycling'],
    'Diet': ['Low-Carb', 'Low-Carb', 'Low-Fat', 'Low-Fat', 'Low-Carb', 'Low-Carb', 'Low-Fat', 'Low-Fat', 'Low-Carb', 'Low-Carb', 'Low-Fat', 'Low-Fat'],
    'Weight_Loss': [5.2, 4.8, 6.1, 5.9, 4.5, 4.7, 6.3, 6.0, 5.0, 5.2, 6.0, 5.8]
}

df = pd.DataFrame(data)

# Fit the model
model = ols('Weight_Loss ~ C(Exercise) * C(Diet)', data=df).fit()

# Perform the ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)

print(anova_table)
```

#### Why Use OLS?
Ordinary Least Squares (OLS) is a method used to estimate the parameters of a linear regression model. It helps determine how the independent variables (factors) relate to the dependent variable (outcome). Here’s why it’s used in this context:

1. **Linear Relationships:** OLS assumes a linear relationship between the predictors (factors) and the outcome. If you believe that Exercise and Diet linearly affect Weight_Loss, OLS is appropriate.

2. **Simplicity and Interpretability:** OLS is straightforward and provides easy-to-interpret results. It gives you coefficients that show how changes in the predictors affect the outcome.

3. **Assumptions of ANOVA:** ANOVA is essentially a special case of OLS where the predictors are categorical. By fitting a model with OLS, you can test the significance of these categorical predictors.

### 3. One-Way ANOVA with Repeated Measures

**One-Way ANOVA with Repeated Measures** is used when the same subjects are measured multiple times under different conditions.

- **Null Hypothesis ($H_0$)**: The means are the same across all conditions (time points, treatments, etc.).
  $$
  H_0: \mu_1 = \mu_2 = \mu_3 = \ldots = \mu_k
  $$
  
- **Alternative Hypothesis ($H_a$)**: At least one condition mean is different.
  $$
  H_a: \text{At least one } \mu_i \text{ is different.}
  $$

**Example:**

You measure the blood pressure of patients before, during, and after a treatment.

- **Measurements**: Blood pressure before, during, and after treatment.
- **Hypotheses**:
  - $H_0$: The mean blood pressure is the same at all time points.
  - $H_a$: The mean blood pressure differs at some time point.

The **one-way ANOVA with repeated measures** tests whether the treatment has a significant effect on blood pressure over time.

### 4. Two-Way ANOVA with Repeated Measures

**Two-Way ANOVA with Repeated Measures** is used when there are two independent variables and at least one involves repeated measures on the same subjects.

- **Null Hypotheses**:
  - $H_0^1$: The means are equal across the levels of the first factor.
  - $H_0^2$: The means are equal across the levels of the second factor.
  - $H_0^3$: There is no interaction between the two factors.

- **Alternative Hypotheses**:
  - $H_a^1$: The means differ across the levels of the first factor.
  - $H_a^2$: The means differ across the levels of the second factor.
  - $H_a^3$: There is an interaction between the two factors.

**Example:**

You are studying the effect of two factors—drug dosage (low, medium, high) and time (morning, afternoon, evening)—on blood pressure in the same group of patients.

- **Factors**: Drug Dosage and Time of Day.
- **Hypotheses**:
  - $H_0^1$: The mean blood pressure is the same for all drug dosages.
  - $H_0^2$: The mean blood pressure is the same at all times of the day.
  - $H_0^3$: There is no interaction between drug dosage and time of day.

The **two-way ANOVA with repeated measures** tests these hypotheses, accounting for the repeated measurements on the same patients.

### Summary of Hypothesis Testing in ANOVA

- **One-Way ANOVA**: Tests if there is a significant difference between the means of three or more groups.
- **Two-Way ANOVA**: Tests the main effects of two factors and their interaction.
- **One-Way ANOVA with Repeated Measures**: Tests for differences when the same subjects are measured under different conditions.
- **Two-Way ANOVA with Repeated Measures**: Tests for the main effects and interaction of two factors with repeated measurements on the same subjects.

ANOVA is a powerful tool for hypothesis testing, allowing you to determine if differences between group means are statistically significant.
